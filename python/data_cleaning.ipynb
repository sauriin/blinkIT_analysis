## Category Standardization
### Cleaning inconsistent categorical values to maintain data quality and ensure accurate analysis.

```python
df['Item_Fat_Content'] = df['Item_Fat_Content'].replace({
    'LF':"Low Fat",
    "low fat":"Low Fat",
    "reg":"Regular"
})
df['Item_Fat_Content'].value_counts()
```

## Duplicate Record Cleaning

### Removing duplicate records to prevent double-counting and incorrect business calculations.

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

## Missing Value Treatment

### Handling missing values in numerical and categorical columns to maintain data quality.

```python
df[['Item_Weight','Item_MRP']].median()
```

```python
df['Item_Weight'] = df['Item_Weight'].fillna(df['Item_Weight'].median())
```

```python
df['Item_MRP'] = df['Item_MRP'].fillna(df['Item_MRP'].median())
```

```python
df[['Item_Weight','Item_MRP']].isnull().sum()
```

## Categorical Missing Value Treatment

### Filling missing values in categorical columns using the most frequent category (mode).

```python
df['Outlet_Size'].isnull().sum()
```

```python
df['Outlet_Size'].mode()
```

```python
df['Outlet_Size'] = df['Outlet_Size'].fillna(
    df['Outlet_Size'].mode()[0]
)
df['Outlet_Size'].isnull().sum()
```

```python
df['Outlet_Size'].value_counts()
```
